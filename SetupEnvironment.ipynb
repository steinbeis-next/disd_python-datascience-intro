# Set up

Notebooks 1–2 use Python. Notebooks 3–4 also use pandas; charts use Matplotlib. The included datasets work offline.

Open a terminal in the repository folder and run the commands below.

Use the commands for your system in a **terminal**, not a notebook cell.

**macOS / Linux**

```bash
python3 -m venv .venv
source .venv/bin/activate
python -m pip install -r requirements.txt
jupyter lab
```

**Windows PowerShell**

```powershell
py -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install -r requirements.txt
jupyter lab
```

Open Notebook 1. Choose the Python 3 kernel if asked.

In JupyterLab, open Notebook 1. If asked, choose the **Python 3** kernel.

## Common problems

- `ModuleNotFoundError`: activate the environment and install `requirements.txt`.
- CSV not found: start JupyterLab in the repository folder; keep `data/` beside the notebooks.
- Strange results: restart the kernel and run cells in order.
- No chart: install requirements; Matplotlib is optional for analysis.
