## Step 1: State the question

## Step 2: Load and inspect the data

Check the rows, fields, and data types. This helps catch a misspelled column or an unexpected value before analysis.

```python
from pathlib import Path
import pandas as pd

data_file = Path("data/04_openaq/air_quality_sample.csv")
df = pd.read_csv(data_file, parse_dates=["date"])
print("Rows:", len(df))
print("Columns:", list(df.columns))
df.head()
```

Use `info()` and `value_counts()` to check the columns and the pollutant labels.

```python
df.info()
df["pollutant"].value_counts()
```

### Check coverage before comparing cities

A group average is easier to interpret when you know how many readings and stations are included. Count rows and stations for each city.

```python
print("Readings by city:")
print(df.groupby("city").size())
print("Stations by city:")
print(df.groupby("city")["station"].nunique())
```

## Step 3: Select PM2.5 and group by city

First select only PM2.5 rows. Then group those rows by city and calculate the mean. Grouping keeps the city comparison explicit.

```python
pm25 = df.loc[df["pollutant"] == "PM2.5"]
city_means = pm25.groupby("city")["value"].mean().round(1)
print("Synthetic PM2.5 sample, mean by city:")
print(city_means)
```

**Try it:** change `"PM2.5"` to `"NO2"`. What changes? Keep the pollutants separate; they measure different things.

## Step 4: Chart the result (optional)

A chart helps communicate the comparison. Label the unit and make it clear that the sample is invented practice data.

```python
import matplotlib.pyplot as plt

ax = city_means.plot.bar(color=["#21867A", "#355C9B"], rot=0)
ax.set_ylabel("Mean synthetic PM2.5 value (µg/m³)")
ax.set_title("Practice data only: not real city measurements")
plt.tight_layout()
plt.show()
```

### Check the result

Confirm that the values are plausible for this tiny sample and that both cities have data. A correct-looking number is not enough: check the selected pollutant, unit, and group sizes.

```python
print("Cities included:", list(city_means.index))
print("Unit in source data: µg/m³")
print("Source: invented workshop sample")
```
